### OBJECTIVE

Create a simple shopping cart program where a customer can view products, add items to a cart, view the bill and
complete checkout.

```python
products = {
    "P1": {"name": "Shoes", "price": 2500, "stock": 5},
    "P2": {"name": "Shirt", "price": 1500, "stock": 10},
    "P3": {"name": "Bag", "price": 3000, "stock": 4},
    "P4": {"name": "Watch", "price": 4000, "stock": 3},
    "P5": {"name": "Headphones", "price": 2000, "stock": 6}
}

cart = {}

while True:
    print("\n" + "=" * 40)
    print("       ONLINE SHOPPING CART MENU")
    print("=" * 40)
    print("1. View Products")
    print("2. Add to Cart")
    print("3. View Cart")
    print("4. Checkout")
    print("5. Exit")
    print("=" * 40)

    choice = input("Enter your choice (1-5): ").strip()

    if choice == "1":
        print("\n--- Available Products ---")
        print(f"{'Product ID':<12} {'Product':<15} {'Price':<15} {'Available Stock':<15}")
        print("-" * 58)
        for pid, details in products.items():
            print(f"{pid:<12} {details['name']:<15} Rs. {details['price']:<11,} {details['stock']:<15}")

    elif choice == "2":
        pid = input("\nEnter Product ID to add: ").strip().upper()

        if pid in products:
            qty_input = input("Enter quantity: ").strip()

            if qty_input.isdigit() and int(qty_input) > 0:
                qty = int(qty_input)
                current_cart_qty = cart.get(pid, 0)
                total_requested = current_cart_qty + qty

                if total_requested <= products[pid]["stock"]:
                    cart[pid] = total_requested
                    print(f"Added {qty} x {products[pid]['name']} to your cart.")
                else:
                    available_to_add = products[pid]["stock"] - current_cart_qty
                    print(f"Insufficient stock! You already have {current_cart_qty} in your cart.")
                    print(f"You can only add up to {available_to_add} more.")
            else:
                print("Invalid quantity. Please enter a positive number.")
        else:
            print("Product ID not found. Please enter P1, P2, P3, P4, or P5.")

    elif choice == "3":
        if not cart:
            print("\nYour cart is currently empty.")
        else:
            print("\n--- Your Shopping Cart ---")
            print(f"{'Product':<15} {'Qty':<8} {'Subtotal':<15}")
            print("-" * 40)
            cart_total = 0

            for pid, qty in cart.items():
                name = products[pid]["name"]
                price = products[pid]["price"]
                subtotal = price * qty
                cart_total += subtotal
                print(f"{name:<15} {qty:<8} Rs. {subtotal:,}")

            print("-" * 40)
            print(f"Cart Total: Rs. {cart_total:,}")

    elif choice == "4":
        if not cart:
            print("\nYour cart is empty. Add items before checking out.")
        else:
            print("\n--- Checkout Summary ---")
            cart_total = 0

            for pid, qty in cart.items():
                name = products[pid]["name"]
                price = products[pid]["price"]
                subtotal = price * qty
                cart_total += subtotal
                print(f"{name} (x{qty}) - Rs. {subtotal:,}")

            print("-" * 35)
            print(f"Subtotal:       Rs. {cart_total:,}")

            if cart_total >= 5000:
                discount = cart_total * 0.10
                final_bill = cart_total - discount
                print(f"Discount (10%): -Rs. {discount:,.2f}")
            else:
                discount = 0
                final_bill = cart_total
                print("Discount:       Rs. 0 (Spend Rs. 5,000+ for 10% off)")

            print(f"Final Bill:     Rs. {final_bill:,.2f}")
            print("-" * 35)

            for pid, qty in cart.items():
                products[pid]["stock"] -= qty

            cart.clear()
            print("Thank you for your purchase! Your order is complete and the cart is now empty.")

    elif choice == "5":
        print("\nThank you for visiting! Goodbye.")
        break

    else:
        print("\nInvalid choice. Please select a number between 1 and 5.")
```
