Q1. Write a Python function count_vowels that accepts a string and returns the number of vowels a, e, i, o, u, A, E, I, O, U in that string.

- Example: 'Welcome'
- Output: 3 vowels

```python
def count_vowels(text: str) -> int:
    vowels = "aeiouAEIOU"
    return sum(1 for char in text if char in vowels)

example = 'Welcome'
output = count_vowels(example)
print(f"Output: {output} vowels")  
```

Q3. Write a Python function display_prices that takes a product name and an arbitrary number of keyword arguments representing product prices. The function should display each product and its price.

Expected Output:

Product: Laptop

Amazon: $800

eBay: $750

Walmart: $780

```python
def display_prices(product_name: str, **prices) -> None:
    print(f"Product: {product_name}\n")
    for store, price in prices.items():
        print(f"{store}: ${price}\n")

display_prices(
    product_name="Laptop", 
    Amazon=800, 
    eBay=750, 
    Walmart=780
)
```

Q4. Write a Python function find_max that takes arbitrary number of positional arguments and returns the maximum number.

```python
def find_max(*args: float) -> float:
    if not args:
        return None
    
    maximum = args[0]
    for num in args:
        if num > maximum:
            maximum = num
    return maximum

result1 = find_max(10, 45, 23, 99, 5)
print(f"Max value: {result1}")  

result2 = find_max(-1.5, -5.2, -0.8)
print(f"Max value: {result2}")  
```

Q6.

Write a Python function calculate_rectangle that accepts the length and width of a rectangle as arguments. Inside this function, create two nested functions:

- calculate_area: This function should calculate the area of the rectangle.
- calculate_perimeter: This function should calculate the perimeter of the rectangle.
- The main function should return both the area and the perimeter of the rectangle.

Expected Output:

Area: 15

Perimeter: 16


```python
def calculate_rectangle(length: float, width: float) -> tuple[float, float]:
    def calculate_area() -> float:
        return length * width

    def calculate_perimeter() -> float:
        return 2 * (length + width)

    area = calculate_area()
    perimeter = calculate_perimeter()
    
    return area, perimeter

rect_area, rect_perimeter = calculate_rectangle(5, 3)

print(f"Area: {rect_area}\n")
print(f"Perimeter: {rect_perimeter}")
```

Q8. Write a nested function to calculate the total hotel bill based on the number of nights.
A hotel charges Rs. 5,000 per night. The customer must also pay 10% tax on the room cost

Expected Output:
Room Cost: 15000
Tax: 1500.0
Total Bill: 16500.0

```python
def calculate_hotel_bill(nights: int) -> float:
    room_rate_per_night = 5000
    room_cost = nights * room_rate_per_night
    
    def calculate_tax(amount: float) -> float:
        return amount * 0.10

    tax = calculate_tax(room_cost)
    total_bill = room_cost + tax
    
    print(f"Room Cost: {room_cost}")
    print(f"Tax: {tax}")
    print(f"Total Bill: {total_bill}")
    
    return total_bill

bill = calculate_hotel_bill(3)
```

Q9.User Login Validation
A website checks a user's username and password before allowing access. Write a nested function to validate the login details

```python
def login_system(username_input: str, password_input: str) -> bool:
    CORRECT_USERNAME = "admin_user"
    CORRECT_PASSWORD = "SecurePassword123"

    def validate_credentials() -> bool:
        is_username_valid = username_input == CORRECT_USERNAME
        is_password_valid = password_input == CORRECT_PASSWORD
        return is_username_valid and is_password_valid

    if validate_credentials():
        print("Access Granted: Welcome back!")
        return True
    else:
        print("Access Denied: Invalid username or password.")
        return False

print("--- Test 1 ---")
login_system("admin_user", "SecurePassword123")

print("\n--- Test 2 ---")
login_system("wrong_user", "password123")
```

Q10.

Write a Python function shopping_bill(price, quantity, discount=0) that uses nested functions:

- An inner function calculate_total() → multiplies price * quantity.
- An inner function apply_discount(total) → subtracts discount from total (but never below 0).

The outer function should return the final bill amount.

Expected Output:

- Final Bill = 600 PKR (no discount)
- Final Bill = 500 PKR (with discount)

```python
def shopping_bill(price: float, quantity: int, discount: float = 0) -> float:
    def calculate_total() -> float:
        return price * quantity

    def apply_discount(total: float) -> float:
        final_amount = total - discount
        return max(0.0, final_amount)

    base_total = calculate_total()
    final_bill = apply_discount(base_total)
    
    return final_bill

# Case 1: No discount
bill_no_discount = shopping_bill(price=200, quantity=3)
print(f"Final Bill = {int(bill_no_discount)} PKR (no discount)")

# Case 2: 100 PKR discount
bill_with_discount = shopping_bill(price=200, quantity=3, discount=100)
print(f"Final Bill = {int(bill_with_discount)} PKR (with discount)")
```
