#### Question 1
#### Employee Data Bank.



Make a class called Employee. The __init__() method for Employee should store its different attributes: name, number, address, salary, post, year of joining, education, and his year of experience.

Make a method called describe_employee() that prints the detail Information about employee, a method update_ year_of_experience(), and a method called increase_salary() that do increamnet in it's salary.

Make an instance called employee101 from your class. Print the four attri-
butes individually, and then call all it's methods for describing employee, ssalary increamnet and increase year of experience.

Create 10 different instances from the class, and call describe_employee() for each
instance. Then do some updations with it's methods.



```python
class Employee:
    def __init__(self, name, number, address, salary, post, year_of_joining, education, year_of_experience):
        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.year_of_joining = year_of_joining
        self.education = education
        self.year_of_experience = year_of_experience

    def describe_employee(self):
        print(f"\n--- Employee Information ---")
        print(f"Name: {self.name}")
        print(f"ID/Number: {self.number}")
        print(f"Address: {self.address}")
        print(f"Salary: ${self.salary:,}")
        print(f"Post/Role: {self.post}")
        print(f"Year of Joining: {self.year_of_joining}")
        print(f"Education: {self.education}")
        print(f"Experience: {self.year_of_experience} years")

    def update_year_of_experience(self, years=1):
        self.year_of_experience += years
        print(f"[Update] {self.name}'s experience updated to {self.year_of_experience} years.")

    def increase_salary(self, amount):
        self.salary += amount
        print(f"[Update] {self.name}'s salary increased by ${amount:,}. New salary: ${self.salary:,}")


print("--- Creating employee101 ---")
employee101 = Employee(
    name="Alice Smith",
    number="EMP101",
    address="123 Maple St, NY",
    salary=75000,
    post="Software Engineer",
    year_of_joining=2022,
    education="B.Sc. Computer Science",
    year_of_experience=4
)

print(f"Name attribute: {employee101.name}")
print(f"Post attribute: {employee101.post}")
print(f"Salary attribute: {employee101.salary}")
print(f"Experience attribute: {employee101.year_of_experience}")

employee101.describe_employee()
employee101.update_year_of_experience(1)
employee101.increase_salary(5000)
employee101.describe_employee()  


print("\n" + "="*40 + "\nGenerating 10 different employee records...\n" + "="*40)

employees = [
    Employee("John Doe", "EMP001", "101 Alpine Rd", 60000, "HR Coordinator", 2023, "BBA", 3),
    Employee("Jane Doe", "EMP002", "102 Alpine Rd", 85000, "Data Scientist", 2021, "M.Sc. Data Science", 5),
    Employee("Bob Johnson", "EMP003", "205 Pine Blvd", 55000, "QA Tester", 2024, "B.Tech", 1),
    Employee("Charlie Brown", "EMP004", "404 Oak Ave", 95000, "DevOps Engineer", 2020, "B.Sc. IT", 7),
    Employee("Diana Prince", "EMP005", "777 Amazon Way", 120000, "Project Manager", 2018, "MBA", 10),
    Employee("Evan Wright", "EMP006", "312 Elm St", 50000, "Support Specialist", 2025, "High School", 0),
    Employee("Fiona Gallagher", "EMP007", "456 South Side", 65000, "Accountant", 2022, "B.Com", 4),
    Employee("George Costanza", "EMP008", "888 Queens Blvd", 70000, "Assistant to the GM", 2023, "BA", 6),
    Employee("Harley Quinn", "EMP009", "901 Arkham Ln", 110000, "Lead Psychologist", 2019, "Ph.D.", 8),
    Employee("Ian Malcolm", "EMP010", "314 Chaos Theory Dr", 105000, "Research Analyst", 2021, "Ph.D. Math", 9)
]

for emp in employees:
    emp.describe_employee()

print("\n" + "="*40 + "\nPerforming updates on employees...\n" + "="*40)

employees[1].increase_salary(7000)

employees[5].update_year_of_experience(1)
employees[5].increase_salary(3000)

employees[1].describe_employee()
employees[5].describe_employee()
```
