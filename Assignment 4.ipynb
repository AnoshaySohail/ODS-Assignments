Q1 Stop sending emails after 3 users (Using for loop and break)

users = ["Ali", "Sara", "John", "Zara", "Mike"]

Output:

Email sent to Ali

Email sent to Sara

Email sent to John

Limit reached. Stop sending emails.

```python
users = ["Ali", "Sara", "John", "Zara", "Mike"]

count = 0

for mail in users:
    print(f"Email sent to {mail}.")
    count += 1
    
    if count == 3:
        print("Limit reached. Stop sending emails.")
        break
```

Q2. Stop monitoring temperature if critical temperature reached (Using for loop and break)

temps = [22, 25, 30, 45, 28] critical = 40

Ouput:

Temperature: 22

Temperature: 25

Temperature: 30

Critical temperature reached! Stop monitoring.

```python
temps =  [22,25, 30, 45, 28]
critical = 40

for temp in temps:
    if temp >= critical:
        print("Critical temperature reached! Stop monitoring.")
        break
        
    print(f"Temperature: {temp}")
```

Q3. Calculate total price of items in a shopping cart (Using For loop)

cart = [100, 250, 50, 300]

Output:

Total Price: 700

```python
cart = [100, 250, 50, 300]
total = 0

for item in cart:
    total += item
    
print(f"Total price: {total}")
```

Q4. Skip customers with unpaid bills (Using For loop and continue)

customers = [ {"name": "Ali", "paid": True},

{"name": "Sara", "paid": False},

{"name": "John", "paid": True}
]

Output:

Access granted to: Ali

Access granted to: John

```python
customers = [ 
    {"name": "Ali", "paid": True},
    {"name": "Sara", "paid": False},
    {"name": "John", "paid": True}
]

for customer in customers:
    if not customer["paid"]:
        continue
        
    print(f"Access granted to: {customer['name']}")
```

Q6. Extract domain names from email list (Using for loop)

emails = ["ali@gmail.com", "sara@yahoo.com", "john@hotmail.com"]

Output:

domains: ['gmail.com', 'yahoo.com', 'hotmail.com']



```python
emails = ["ali@gmail.com", "sara@yahoo.com", "john@hotmail.com"]
domains = []

for email in emails:
    domain_name = email.split("@")[1]
    domains.append(domain_name)

print(f"domains: {domains}")
```

Q7. Create a list of numbers, generate a new list containing only the numbers that are divisible by 3 or 5. (List Comprehension)

```python
numbers = [14, 42, 78, 5, 91, 23, 66, 49, 87, 31]

filtered_numbers = [num for num in numbers if num % 3 == 0 or num % 5 == 0]

print(f"Filtered list: {filtered_numbers}")
```

Q8. Create a multiplication table using a nested for loop.

numbers = [2, 3, 4]
multipliers = [1, 2, 3]

Expected Output:

2 x 1 = 2
2 x 2 = 4
2 x 3 = 6
3 x 1 = 3
3 x 2 = 6
3 x 3 = 9
4 x 1 = 4
4 x 2 = 8
4 x 3 = 12

Use a nested for loop.

```python
numbers = [2, 3, 4] 
multipliers = [1, 2, 3]

for num in numbers:
    for mult in multipliers:
        table = num * mult
        print(f"{num} x {mult} = {table}")
```

Q9. Keep asking the user for a password until the correct password is entered.

correct_password = "python123"

Expected behaviour:

Enter password: abc
Incorrect password. Try again.

Enter password: 12345
Incorrect password. Try again.

Enter password: python123
Access granted!

Use a while loop.

```python
correct_password = "python123"

while True:
    user_input = input("Enter password: ")
    print(f"User input: {user_input}")

    if user_input == correct_password:
        print("Access granted!")
        break
    else:
        print("Incorrect password. Try again.\n")
```

Q10. Count down from 5 to 1 and then display "Time's up!".

Expected Output:

5
4
3
2
1
Time's up!

Use a while loop.

```python
count = 5

while count > 0:
    print(count)

    count -= 1
print("Time's up!")
```
